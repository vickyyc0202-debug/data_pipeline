```python
# ============================================================
# DATA PROCESSING + MODELING PIPELINE
# Suitable for timed data projects
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, cross_val_score, KFold, TimeSeriesSplit
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

# ============================================================
# 0. CONFIG
# ============================================================

FILE_PATH = "data.csv"

TARGET = "price"

# Set to None if not time-series
TIME_COL = "datetime"

# Optional ID column
ID_COL = None

# Columns you know should be removed
DROP_COLS = []

TEST_SIZE = 0.20

RANDOM_STATE = 42

# Whether to create lag / rolling target features
USE_TIME_FEATURES = True

# Whether to clip extreme values
USE_OUTLIER_CLIP = True


# ============================================================
# 1. LOAD DATA
# ============================================================

df = pd.read_csv(FILE_PATH)

# If Excel:
# df = pd.read_excel(FILE_PATH)

print("\n===== BASIC DATA INFO =====")
print("Shape:", df.shape)
print(df.head())
print(df.info())


# ============================================================
# 2. BASIC CLEANING
# ============================================================

# Drop duplicates
df = df.drop_duplicates()

# Drop unnecessary columns
df = df.drop(columns=DROP_COLS, errors="ignore")

# Optional ID
if ID_COL is not None and ID_COL in df.columns:
    df = df.set_index(ID_COL)

# Parse datetime
if TIME_COL is not None and TIME_COL in df.columns:
    df[TIME_COL] = pd.to_datetime(
        df[TIME_COL],
        errors="coerce"
    )

    df = df.sort_values(TIME_COL).reset_index(drop=True)


# ============================================================
# 3. EDA
# ============================================================

print("\n===== SUMMARY STATISTICS =====")
print(df.describe(include="all").T)

print("\n===== MISSING VALUES =====")
missing = df.isna().sum()
print(
    missing[missing > 0]
    .sort_values(ascending=False)
)

print("\n===== TARGET =====")
print(df[TARGET].describe())

# Regression target distribution
print("Target skewness:", df[TARGET].skew())

# Numeric correlations with target
numeric_df = df.select_dtypes(include=np.number)

if TARGET in numeric_df.columns:
    corr = (
        numeric_df
        .corr()[TARGET]
        .sort_values(ascending=False)
    )

    print("\n===== CORRELATION WITH TARGET =====")
    print(corr)


# ============================================================
# 4. TIME FEATURE ENGINEERING
# ============================================================

if TIME_COL is not None and TIME_COL in df.columns:

    dt = df[TIME_COL]

    # ----------------------------
    # Basic time features
    # ----------------------------

    df["hour"] = dt.dt.hour
    df["day_of_week"] = dt.dt.dayofweek
    df["day"] = dt.dt.day
    df["month"] = dt.dt.month
    df["quarter"] = dt.dt.quarter
    df["year"] = dt.dt.year

    df["is_weekend"] = (
        df["day_of_week"] >= 5
    ).astype(int)


    # ----------------------------
    # Optional cyclical encoding
    # ----------------------------

    df["hour_sin"] = np.sin(
        2 * np.pi * df["hour"] / 24
    )

    df["hour_cos"] = np.cos(
        2 * np.pi * df["hour"] / 24
    )

    df["dow_sin"] = np.sin(
        2 * np.pi * df["day_of_week"] / 7
    )

    df["dow_cos"] = np.cos(
        2 * np.pi * df["day_of_week"] / 7
    )

    df["month_sin"] = np.sin(
        2 * np.pi * df["month"] / 12
    )

    df["month_cos"] = np.cos(
        2 * np.pi * df["month"] / 12
    )


# ============================================================
# 5. LAG / ROLLING FEATURES
# ============================================================

if (
    USE_TIME_FEATURES
    and TIME_COL is not None
    and TIME_COL in df.columns
):

    # IMPORTANT:
    # shift first, then rolling
    # so current target does not leak into features

    for lag in [1, 2, 3, 7]:

        df[f"{TARGET}_lag_{lag}"] = (
            df[TARGET]
            .shift(lag)
        )


    for window in [7, 14, 30]:

        past_target = df[TARGET].shift(1)

        df[f"{TARGET}_mean_{window}"] = (
            past_target
            .rolling(window)
            .mean()
        )

        df[f"{TARGET}_std_{window}"] = (
            past_target
            .rolling(window)
            .std()
        )

        df[f"{TARGET}_min_{window}"] = (
            past_target
            .rolling(window)
            .min()
        )

        df[f"{TARGET}_max_{window}"] = (
            past_target
            .rolling(window)
            .max()
        )


    # Previous-period percentage change
    df[f"{TARGET}_pct_change"] = (
        df[TARGET]
        .pct_change()
        .shift(1)
    )


# ============================================================
# 6. OPTIONAL DOMAIN FEATURES
# ============================================================

# Only add interactions that make sense.

# Example:
#
# if "demand" in df.columns and "capacity" in df.columns:
#     df["demand_capacity_ratio"] = (
#         df["demand"]
#         / df["capacity"].replace(0, np.nan)
#     )
#
# if "income" in df.columns and "population" in df.columns:
#     df["income_per_capita"] = (
#         df["income"]
#         / df["population"].replace(0, np.nan)
#     )


# ============================================================
# 7. REMOVE INVALID TARGET ROWS
# ============================================================

# Never impute the target unless there is a strong reason.
df = df.dropna(subset=[TARGET])


# ============================================================
# 8. DEFINE X / y
# ============================================================

drop_for_X = [TARGET]

if TIME_COL is not None:
    drop_for_X.append(TIME_COL)

X = df.drop(
    columns=drop_for_X,
    errors="ignore"
)

y = df[TARGET]


# ============================================================
# 9. TRAIN / TEST SPLIT
# ============================================================

if TIME_COL is not None and TIME_COL in df.columns:

    # TIME SERIES:
    # past = train
    # future = test

    split_idx = int(
        len(df) * (1 - TEST_SIZE)
    )

    X_train = X.iloc[:split_idx].copy()
    X_test = X.iloc[split_idx:].copy()

    y_train = y.iloc[:split_idx].copy()
    y_test = y.iloc[split_idx:].copy()

else:

    # NON-TIME-SERIES

    X_train, X_test, y_train, y_test = (
        train_test_split(
            X,
            y,
            test_size=TEST_SIZE,
            random_state=RANDOM_STATE
        )
    )


print("\n===== DATA SPLIT =====")
print("Train:", X_train.shape)
print("Test :", X_test.shape)


# ============================================================
# 10. IDENTIFY NUMERIC / CATEGORICAL COLUMNS
# ============================================================

numeric_cols = (
    X_train
    .select_dtypes(include=np.number)
    .columns
    .tolist()
)

categorical_cols = (
    X_train
    .select_dtypes(
        include=["object", "category", "bool"]
    )
    .columns
    .tolist()
)

print("\nNumeric columns:")
print(numeric_cols)

print("\nCategorical columns:")
print(categorical_cols)


# ============================================================
# 11. MISSING VALUES
# ============================================================

# ----------------------------
# Numeric:
# train median only
# ----------------------------

for col in numeric_cols:

    median = X_train[col].median()

    X_train[col] = (
        X_train[col]
        .fillna(median)
    )

    X_test[col] = (
        X_test[col]
        .fillna(median)
    )


# ----------------------------
# Categorical:
# train mode only
# ----------------------------

for col in categorical_cols:

    mode = X_train[col].mode()

    if len(mode) > 0:
        fill_value = mode.iloc[0]
    else:
        fill_value = "Unknown"

    X_train[col] = (
        X_train[col]
        .fillna(fill_value)
    )

    X_test[col] = (
        X_test[col]
        .fillna(fill_value)
    )


# ============================================================
# 12. OUTLIER HANDLING
# ============================================================

if USE_OUTLIER_CLIP:

    for col in numeric_cols:

        # thresholds learned from TRAIN only
        lower = X_train[col].quantile(0.01)
        upper = X_train[col].quantile(0.99)

        X_train[col] = (
            X_train[col]
            .clip(lower, upper)
        )

        X_test[col] = (
            X_test[col]
            .clip(lower, upper)
        )


# ============================================================
# 13. CATEGORICAL ENCODING
# ============================================================

# Combine temporarily so dummy columns line up.
#
# This is okay because pd.get_dummies only determines
# category column structure.
#
# For strict production ML, use OneHotEncoder fitted on train.

combined = pd.concat(
    [X_train, X_test],
    axis=0
)

combined = pd.get_dummies(
    combined,
    columns=categorical_cols,
    drop_first=True
)

X_train = combined.iloc[
    :len(X_train)
].copy()

X_test = combined.iloc[
    len(X_train):
].copy()


# Convert bools if necessary
X_train = X_train.astype(float)
X_test = X_test.astype(float)


# ============================================================
# 14. FINAL MISSING CHECK
# ============================================================

# Useful after lag / rolling / ratio features

X_train = X_train.replace(
    [np.inf, -np.inf],
    np.nan
)

X_test = X_test.replace(
    [np.inf, -np.inf],
    np.nan
)

# Re-fill any remaining numeric NaNs
train_medians = X_train.median()

X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)


print("\nRemaining train NaNs:")
print(X_train.isna().sum().sum())

print("Remaining test NaNs:")
print(X_test.isna().sum().sum())


# ============================================================
# 15. SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)


# ============================================================
# 16. BASELINE MODELS
# ============================================================

models = {

    "Linear":
        LinearRegression(),

    "Ridge":
        Ridge(alpha=1.0),

    "Lasso":
        Lasso(
            alpha=0.01,
            max_iter=10000
        ),

    "RandomForest":
        RandomForestRegressor(
            n_estimators=200,
            max_depth=None,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
}


# ============================================================
# 17. TRAIN + EVALUATE
# ============================================================

results = []

trained_models = {}

for name, model in models.items():

    # Trees do not require scaling
    if name == "RandomForest":

        model.fit(
            X_train,
            y_train
        )

        train_pred = model.predict(
            X_train
        )

        test_pred = model.predict(
            X_test
        )

    else:

        model.fit(
            X_train_scaled,
            y_train
        )

        train_pred = model.predict(
            X_train_scaled
        )

        test_pred = model.predict(
            X_test_scaled
        )


    # Metrics
    train_rmse = np.sqrt(
        mean_squared_error(
            y_train,
            train_pred
        )
    )

    test_rmse = np.sqrt(
        mean_squared_error(
            y_test,
            test_pred
        )
    )

    test_mae = mean_absolute_error(
        y_test,
        test_pred
    )

    test_r2 = r2_score(
        y_test,
        test_pred
    )


    results.append({

        "Model": name,

        "Train_RMSE":
            train_rmse,

        "Test_RMSE":
            test_rmse,

        "Test_MAE":
            test_mae,

        "Test_R2":
            test_r2,

        "Overfit_Ratio":
            test_rmse / train_rmse
            if train_rmse > 0
            else np.nan
    })

    trained_models[name] = model


results_df = (
    pd.DataFrame(results)
    .sort_values("Test_RMSE")
)

print("\n===== MODEL RESULTS =====")
print(results_df)


# ============================================================
# 18. SELECT BEST MODEL
# ============================================================

best_model_name = (
    results_df
    .iloc[0]["Model"]
)

best_model = trained_models[
    best_model_name
]

print(
    "\nBest model:",
    best_model_name
)


# ============================================================
# 19. FINAL PREDICTIONS
# ============================================================

if best_model_name == "RandomForest":

    predictions = best_model.predict(
        X_test
    )

else:

    predictions = best_model.predict(
        X_test_scaled
    )


# ============================================================
# 20. FEATURE IMPORTANCE
# ============================================================

if hasattr(best_model, "coef_"):

    importance = pd.Series(

        best_model.coef_,

        index=X_train.columns

    )

    feature_importance = pd.DataFrame({

        "feature":
            importance.index,

        "coefficient":
            importance.values,

        "importance":
            np.abs(
                importance.values
            )
    })

    feature_importance = (
        feature_importance
        .sort_values(
            "importance",
            ascending=False
        )
    )

    print(
        "\n===== TOP FEATURES ====="
    )

    print(
        feature_importance.head(15)
    )


elif hasattr(
    best_model,
    "feature_importances_"
):

    feature_importance = pd.DataFrame({

        "feature":
            X_train.columns,

        "importance":
            best_model.feature_importances_
    })

    feature_importance = (
        feature_importance
        .sort_values(
            "importance",
            ascending=False
        )
    )

    print(
        "\n===== TOP FEATURES ====="
    )

    print(
        feature_importance.head(15)
    )

else:

    feature_importance = None


# ============================================================
# 21. RESIDUAL ANALYSIS
# ============================================================

residuals = (
    y_test.values
    - predictions
)

print(
    "\n===== RESIDUAL CHECK ====="
)

print(
    "Residual mean:",
    residuals.mean()
)

print(
    "Residual std:",
    residuals.std()
)


# ============================================================
# 22. SANITY CHECKS
# ============================================================

print(
    "\n===== SANITY CHECKS ====="
)

print(
    "Actual range:",
    y_test.min(),
    "to",
    y_test.max()
)

print(
    "Prediction range:",
    predictions.min(),
    "to",
    predictions.max()
)

print(
    "Prediction mean:",
    predictions.mean()
)

print(
    "Actual mean:",
    y_test.mean()
)


# ============================================================
# 23. SIMPLE VISUALIZATION
# ============================================================

plt.figure(figsize=(7, 5))

plt.scatter(
    y_test,
    predictions,
    alpha=0.5
)

lims = [
    min(
        y_test.min(),
        predictions.min()
    ),
    max(
        y_test.max(),
        predictions.max()
    )
]

plt.plot(
    lims,
    lims,
    "--"
)

plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.title(
    f"Actual vs Predicted - {best_model_name}"
)

plt.show()


# Residual plot

plt.figure(figsize=(7, 5))

plt.scatter(
    predictions,
    residuals,
    alpha=0.5
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("Predicted")
plt.ylabel("Residual")
plt.title("Residual Plot")

plt.show()


# ============================================================
# 24. OPTIONAL CROSS VALIDATION
# ============================================================

# Only run this if you have enough time.

if best_model_name != "RandomForest":

    cv_model = Ridge(alpha=1.0)

    if TIME_COL is not None:

        cv = TimeSeriesSplit(
            n_splits=5
        )

    else:

        cv = KFold(
            n_splits=5,
            shuffle=True,
            random_state=RANDOM_STATE
        )

    cv_scores = cross_val_score(

        cv_model,

        X_train_scaled,

        y_train,

        cv=cv,

        scoring="neg_root_mean_squared_error"
    )

    print(
        "\n===== CV ====="
    )

    print(
        "CV RMSE:",
        -cv_scores.mean()
    )

    print(
        "CV RMSE std:",
        cv_scores.std()
    )


# ============================================================
# 25. OPTIONAL RIDGE SENSITIVITY
# ============================================================

print(
    "\n===== RIDGE SENSITIVITY ====="
)

for alpha in [
    0.01,
    0.1,
    1,
    10,
    100
]:

    model = Ridge(
        alpha=alpha
    )

    model.fit(
        X_train_scaled,
        y_train
    )

    pred = model.predict(
        X_test_scaled
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            pred
        )
    )

    print(
        f"alpha={alpha:<6} "
        f"RMSE={rmse:.4f}"
    )
```
